```python
# Setup
import sys
sys.path.append('../')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.config import config
from src.data.preprocessing import load_raw_data, clean_data
from src.visualization.visualize import generate_eda_report

# Display settings
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 200)
plt.style.use('seaborn-v0_8-whitegrid')

print("Environment ready")
```

```python
# Load raw data
df_raw = load_raw_data()
df_raw.head()
```

```python
# Basic info
print(f"Shape: {df_raw.shape}")
print(f"\nColumns: {list(df_raw.columns)}")
print(f"\nDtypes:\n{df_raw.dtypes}")
print(f"\nMissing values:\n{df_raw.isnull().sum()}")
print(f"\nDuplicate rows: {df_raw.duplicated().sum()}")
```

```python
# Statistical summary
df_raw.describe(include='all').T
```

```python
# Clean data
df_clean = clean_data(df_raw)

print(f"Cleaned shape: {df_clean.shape}")
print(f"\nColumns: {list(df_clean.columns)}")
print(f"\nMissing after cleaning:\n{df_clean.isnull().sum()}")
df_clean.head()
```

```python
# Target variable analysis
target = config.get('data.target_column', 'selling_price')

print(f"Target: {target}")
print(f"Mean: {df_clean[target].mean():.0f}")
print(f"Median: {df_clean[target].median():.0f}")
print(f"Std: {df_clean[target].std():.0f}")
print(f"Min: {df_clean[target].min():.0f}")
print(f"Max: {df_clean[target].max():.0f}")
print(f"Skewness: {df_clean[target].skew():.3f}")
```

```python
# Generate full EDA report
figures = generate_eda_report(df_clean, target_col=target)

print("EDA report generated. Check reports/figures/")
```
